```python
sebastian_file2 = r'/Users/jeannettearjona/Downloads/osm_no_principales_atributos/osm_no_principales_atributos.shp'
red_con_att_notmain = gpd.read_file(sebastian_file2)
```

```python
red_con_att_notmain[red_con_att_notmain['cap_final'].notna()]['highway'].value_counts()
```

```python
resumen = (
    red_con_att_notmain
    .groupby('highway')
    .agg(
        cap_final_mean=('cap_final', 'mean'),
        cap_final_median=('cap_final', 'median'),
        carr_final_mean=('carr_final', 'mean'),
        carr_final_median=('carr_final', 'median'),
        velprom_mean=('velprom', 'mean'),
        velprom_median=('velprom', 'median'),
        vel_final_mean=('vel_final', 'mean'),
        vel_final_median=('vel_final', 'median'),
    )
    .reset_index()
)
resumen
```
