```python
links['FROM_LON_deg'] = links['FROM_LONGI'] / 1e6
links['FROM_LAT_deg'] = links['FROM_LATIT'] / 1e6
links['TO_LON_deg']   = links['TO_LONGITU'] / 1e6
links['TO_LAT_deg']   = links['TO_LATITUD'] / 1e6
```

```python
from_points = gpd.GeoDataFrame(
    links[['ID']].copy(),
    geometry=gpd.points_from_xy(
        links['FROM_LON_deg'],
        links['FROM_LAT_deg']
    ),
    crs="EPSG:4326"
)

to_points = gpd.GeoDataFrame(
    links[['ID']].copy(),
    geometry=gpd.points_from_xy(
        links['TO_LON_deg'],
        links['TO_LAT_deg']
    ),
    crs="EPSG:4326"
)
```

```python
from_points = from_points.to_crs(links.crs)
to_points   = to_points.to_crs(links.crs)
```

```python
from_points.iloc[:5].explore()
```

```python
from_join = gpd.sjoin_nearest(
    from_points,
    nodes[['NODE_ID','geometry']],
    how='left',
    distance_col='dist_from'
)

to_join = gpd.sjoin_nearest(
    to_points,
    nodes[['NODE_ID','geometry']],
    how='left',
    distance_col='dist_to'
)
```

```python
# --- 4. Pasar resultados a links ---
links = links.merge(
    from_join[['ID', 'NODE_ID']].rename(columns={'NODE_ID': 'fromNode'}),
    on='ID',
    how='left'
)

links = links.merge(
    to_join[['ID', 'NODE_ID']].rename(columns={'NODE_ID': 'toNode'}),
    on='ID',
    how='left'
)
```

```python
links.to_file("links_cleaned.shp")
```
