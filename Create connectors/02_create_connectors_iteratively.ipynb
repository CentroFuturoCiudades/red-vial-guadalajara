### __Filter Active Nodes__

```python
def filter_nodes_by_priority(Visum, priority):
    node_filter = Visum.Filters.NodeFilter()
    node_filter.Init()

    node_filter.AddCondition(
        Op="OP_NONE",
        Complement=False,
        AttID="HIGHWAY_PRIORITY",
        CompareOperator=9, #9 is =
        Val=priority,
        Position=0
    )

    return node_filter
```

```python
# Apply filter
filter_nodes_by_priority(Visum, 1)
```
