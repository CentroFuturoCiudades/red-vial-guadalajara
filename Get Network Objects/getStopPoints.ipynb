```python
import win32com.client as com
import os
import pandas as pd
import geopandas as gpd
from shapely import wkt
```

```python
folder = r'C:\Users\Roberto Ponce López\OneDrive - Instituto Tecnologico y de Estudios Superiores de Monterrey (1)\Modelación Urbana - Red Vial Guadalajara'
visum_network = os.path.join(folder, "Red Base GDL", "RedBase Conectores y Atts", "RedBase 250826.ver")

Visum = com.Dispatch("Visum.Visum")
Visum.LoadVersion(visum_network)
C = com.constants
```

```python
nodes = pd.DataFrame({
    "No": [i[1] for i in Visum.Net.Nodes.GetMultiAttValues("No")],
    "XCoord": [i[1] for i in Visum.Net.Nodes.GetMultiAttValues("XCoord")],
    "YCoord": [i[1] for i in Visum.Net.Nodes.GetMultiAttValues("YCoord")],
    "N_SP": [i[1] for i in Visum.Net.Nodes.GetMultiAttValues(r"COUNT:STOPPOINTS")],
    "INLINK_HAS_SP": [i[1] for i in Visum.Net.Nodes.GetMultiAttValues(r"Distinct:InLinks\COUNT:STOPPOINTS")],
    "OUTLINK_HAS_SP": [i[1] for i in Visum.Net.Nodes.GetMultiAttValues(r"Distinct:OutLinks\COUNT:STOPPOINTS")],
})

nodes = gpd.GeoDataFrame(
    nodes,
    geometry= gpd.points_from_xy(nodes["XCoord"], nodes["YCoord"]),
    crs="EPSG:4326"
)
```
