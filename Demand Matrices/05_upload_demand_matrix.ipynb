### Create demand array for Visum

```python
matrix_ids = {
    "HEAVY_FREIGHT_AP": {"period": 1,  "pf": 2},
    "HEAVY_FREIGHT_LM": {"period": 3, "pf": 4},
    "HEAVY_FREIGHT_EA": {"period": 5,  "pf": 6},
    "HEAVY_FREIGHT_PP": {"period": 7, "pf": 8},

    "LIGHT_FREIGHT_AP": {"period": 9, "pf": 10},
    "LIGHT_FREIGHT_LM": {"period": 11, "pf": 12},
    "LIGHT_FREIGHT_EA": {"period": 13, "pf": 14},
    "LIGHT_FREIGHT_PP": {"period": 15, "pf": 16},

    "VP_EM": {"period": 17, "pf": 18},
    "VP_AP": {"period": 19, "pf": 20},
    "VP_LM": {"period": 21, "pf": 22},
    "VP_EA": {"period": 23, "pf": 24},
    "VP_PP": {"period": 25, "pf": 26},
    "VP_LN": {"period": 27, "pf": 28},

    "TP_EM": {"period": 29, "pf": 30},
    "TP_AP": {"period": 31, "pf": 32},
    "TP_LM": {"period": 33, "pf": 34},
    "TP_EA": {"period": 35, "pf": 36},
    "TP_LN": {"period": 37, "pf": 38},
    "TP_PP": {"period": 39, "pf": 40},
}
```

```python
def create_demand_array(demand_matrix, trips_col, origin_col="Origen_ageb", destination_col="Destino_ageb"):

    demand_array = np.zeros((n_zones, n_zones), dtype=np.float64)

    # indices de origen y destino en la matriz
    origin_indices = (
        demand_matrix[origin_col]
        .map(zone_to_index)
        .to_numpy(dtype=np.int64)
    )

    destination_indices = (
        demand_matrix[destination_col]
        .map(zone_to_index)
        .to_numpy(dtype=np.int64)
    )

    # Numero de viajes
    demand_values = (
        demand_matrix[trips_col]
        .to_numpy(dtype=np.float64)
    )

    # Llenar el array con los viajes reales
    np.add.at(
        demand_array,
        (origin_indices, destination_indices),
        demand_values
    )

    return demand_array
```

```python
DEMAND_FOLDER = os.path.join(folder, "Matriz Origen Destino", "Demanda x PF")

for file_name in sorted(os.listdir(DEMAND_FOLDER)):

    if not file_name.endswith(".csv"):
        continue

    file_path = os.path.join(DEMAND_FOLDER, file_name)

    try:

        matrix_key = (
            os.path.splitext(file_name)[0]
            .removeprefix("matriz_")
        )

        if matrix_key not in matrix_ids:
            print(f"⚠️ {file_name}: no tiene Matrix IDs definidos")
            continue

        # Leer CSV
        df = pd.read_csv(file_path)
        df = df.dropna(subset=["Origen_ageb", "Destino_ageb"]).copy()

        # Crear las dos matrices
        demand_array = create_demand_array(df, "Ponderador")
        demand_array_pf = create_demand_array(df, "Ponderador_x_PF")

        # Obtener Matrix IDs
        period_matrix_no = matrix_ids[matrix_key]["period"]
        pf_matrix_no = matrix_ids[matrix_key]["pf"]

        #visum_matrix = Visum.Net.Matrices.ItemByKey(period_matrix_no)
        #visum_matrix.SetValues(demand_array, Add=False)

        visum_matrix_pf = Visum.Net.Matrices.ItemByKey(pf_matrix_no)
        visum_matrix_pf.SetValues(demand_array_pf, Add=False)

        print(
            f"{file_name}: "
            f"{demand_array.sum():,.2f} viajes | "
            f"{demand_array_pf.sum():,.2f} viajes PF"
        )

    except Exception as e:
        print(f"Could not insertto visum matrix for {file_name}: {e}")

    
```
