```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
accesos_carreteros = {
    "999990001": 10001, # Acceso vallarta
    "999990002": 10002, # Acceso_Colotlan
    "999990003": 10003, # Acceso_Saltillo
    "999990004": 10004, # Acceso_Zapotlanejo_Cuota
    "999990005": 10005, # Acceso_Chapala
    "999990006": 10006, # Acceso_Lopez_Mateos
    "999990007": 10007,   # Acceso_zapotlanejo_Libre
    "99999000A": 5241 # AGEB DE AEROPUERTO
}
```

```python
folder = r'/Users/jeannettearjona/Library/CloudStorage/OneDrive-InstitutoTecnologicoydeEstudiosSuperioresdeMonterrey/Modelación Urbana - Red Vial Guadalajara/Matriz Origen Destino/matrices periodos PF last'
carpeta = Path(folder)

for archivo in carpeta.glob("*.csv"):
    df = pd.read_csv(
        archivo,
        dtype={
            "Origen": "string",
            "Destino": "string",
            "Origen_ageb": "string",
            "Destino_ageb": "string",
        }
    )

    df["Origen"] = df["Origen"].str.strip()
    df["Destino"] = df["Destino"].str.strip()

    # Corregir Origen_ageb según Origen
    mask_origen = df["Origen"].isin(accesos_carreteros)

    df.loc[mask_origen, "Origen_ageb"] = (
        df.loc[mask_origen, "Origen"]
        .map(accesos_carreteros)
        .astype("string")
    )

    # Corregir Destino_ageb según Destino
    mask_destino = df["Destino"].isin(accesos_carreteros)

    df.loc[mask_destino, "Destino_ageb"] = (
        df.loc[mask_destino, "Destino"]
        .map(accesos_carreteros)
        .astype("string")
    )

    # Sobrescribir el archivo original
    df.to_csv(
        archivo,
        index=False,
        encoding="utf-8-sig"
    )

    print(
        f"{archivo.name}: "
        f"{mask_origen.sum()} orígenes y "
        f"{mask_destino.sum()} destinos corregidos"
    )
```
