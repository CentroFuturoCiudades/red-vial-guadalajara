```python
import pandas as pd
import numpy as np
import os
```

```python
PATH = r"/Users/jeannettearjona/Library/CloudStorage/OneDrive-InstitutoTecnologicoydeEstudiosSuperioresdeMonterrey/Modelación Urbana - Red Vial Guadalajara"
PF_FOLDER = "Matriz Origen Destino/Matrices por periodo-modo + peak factor ponderador"
DEMAND_FOLDER = "Matriz Origen Destino/Matrices filtradas por periodo y modo (con accesos a carretera)"
OUT_PATH = "Matriz Origen Destino/Demanda x PF"
```

#### Read PFs

```python
peak_factors = pd.read_csv(os.path.join(PATH, PF_FOLDER, "diccionario_peak_factors_v2.csv"), index_col=0)
peak_factors
```

```python
for file in sorted(os.listdir(os.path.join(PATH, DEMAND_FOLDER))):
    if not file.lower().endswith(".csv"):
        continue

    matriz = pd.read_csv(os.path.join(PATH, DEMAND_FOLDER, file))

    matriz = matriz.merge(
        peak_factors,
        how="left",
        on="Periodo",
    )

    matriz = matriz.rename(columns={
        "Sum of Ponderador": "Ponderador",
        "Peak factor por periodo": "Peak Factor"
    })

    matriz["Ponderador_x_PF"] = (
        matriz["Ponderador"] * matriz["Peak Factor"]
    )

    matriz.to_csv(
        os.path.join(PATH, OUT_PATH, file),
        index=False
    )

    print(f"Guardado: {file}")
```
