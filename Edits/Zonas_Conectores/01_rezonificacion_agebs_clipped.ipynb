```python
import pandas as pd
import geopandas as gpd
import numpy as np
import os
```

```python
FOLDER = r"C:\Users\Roberto Ponce López\OneDrive - Instituto Tecnologico y de Estudios Superiores de Monterrey (1)\Modelación Urbana - Red Vial Guadalajara"
ZONAS_FOLDER = "red_shapefiles\zones_agebs_visum"
```

```python
# 2,210 zonas que están en Visum
visum_zones = gpd.read_file(os.path.join(FOLDER, ZONAS_FOLDER, "con accesos carreteros", "new_zones_with_access.shp"))

# 44 zonas (agebs_clipped) que Gonzalo arreglo
rural_agebs_clipped = gpd.read_file(os.path.join(FOLDER, ZONAS_FOLDER, "rezonificacion", "rural_agebs_clipped.gpkg"), layer="rural_agebs_clipped")

# pasar visum_id a las clipped zones
rural_agebs_clipped = rural_agebs_clipped.merge(
    visum_zones[["clave_ageb", "id_visum"]],
    left_on="CVEGEO",
    right_on="clave_ageb",
    how="left"
).drop(columns="clave_ageb")

rural_agebs_clipped.to_file(os.path.join(FOLDER, ZONAS_FOLDER, "rezonificacion", "rural_agebs_clipped.gpkg"), layer="rural_agebs_clipped")
```
