```python
import pandas as pd
import geopandas as gpd
import ast
```

## Links viales (SHP)

```python
drive_links = gpd.read_file(r'/Users/jeannettearjona/Library/CloudStorage/OneDrive-InstitutoTecnologicoydeEstudiosSuperioresdeMonterrey/Modelación Urbana - Red Vial Guadalajara/Pruebas OSM-OSMNX/osmnx-AMG/Highway/Edges/highway_edges.shp')
all_links = gpd.read_file(r'/Users/jeannettearjona/Library/CloudStorage/OneDrive-InstitutoTecnologicoydeEstudiosSuperioresdeMonterrey/Modelación Urbana - Red Vial Guadalajara/Pruebas OSM-OSMNX/osmnx-AMG/Highway_All/Edges/all_edges.shp')
```

```python
def clean_highway(x):
    if isinstance(x, str) and x.strip().startswith('['):
        try:
            # convierte string -> lista real
            lst = ast.literal_eval(x)
            return ";".join(lst)
        except:
            return x
    return x
```

```python
drive_links['highway_list'] = drive_links['highway'].apply(clean_highway)
all_links['highway_list'] = all_links['highway'].apply(clean_highway)
```

```python
highway_unique_drive = drive_links['highway_list'].unique()
highway_unique_all = all_links['highway_list'].unique()
diff = set(highway_unique_all) - set(highway_unique_drive)
diff
```

## Clean Highway column 
- change to tuple

```python
def to_list(x):
    if isinstance(x, str):
        if x.startswith('[') and x.endswith(']'):
            return ast.literal_eval(x)
        else:
            return [x]
    return x
```

```python
drive_links['highway'] = drive_links['highway'].apply(to_list)
drive_links['highway'] = drive_links['highway'].apply(tuple)
drive_links['highway'].unique()
```
